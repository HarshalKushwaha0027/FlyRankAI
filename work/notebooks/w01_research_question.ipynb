## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.

import os

# 1. Clone your GitHub repository into this Colab session
if not os.path.isdir("FlyRankAI"):
    !git clone https://github.com/HarshalKushwaha0027/FlyRankAI

# 2. Change the working directory into the folder so Pandas can find 'data/raw/...'
os.chdir("FlyRankAI")
print("Working directory is now:", os.getcwd())

import pandas as pd

# Load the starter dataset
df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

# Filter out completely dead pages (we only care about pages with actual data)
active_pages = df[df['impressions_90d'] > 100]
total_active = len(active_pages)

# 1. How many active pages are showing a downward trend?
declining_pages = active_pages[active_pages['trend_direction'] == 'down']
decline_pct = (len(declining_pages) / total_active) * 100

# 2. Risk Exposure: How many of these declining pages have HIGH traffic to lose?
# Let's say high traffic is > 5000 impressions in 90 days
high_risk_pages = declining_pages[declining_pages['impressions_90d'] > 5000]
high_risk_pct = (len(high_risk_pages) / len(declining_pages)) * 100

print(f"Total active pages (100+ impressions): {total_active}")
print(f"1. {decline_pct:.1f}% of active pages are already showing a downward trend.")
print(f"2. Of those declining pages, {high_risk_pct:.1f}% have over 5,000 recent impressions.")
print(f"Conclusion: There is massive risk exposure here. Predicting this decline *before* it hits these high-traffic pages is highly valuable.")
```
