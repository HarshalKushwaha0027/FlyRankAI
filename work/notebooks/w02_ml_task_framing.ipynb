## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A fixed rule (like "flag if impressions < 100") is backward-looking; it only catches pages after they have already failed. Predicting future decline requires weighing multiple competing factors simultaneously (e.g., high search volume + decaying CTR + older content age + high competition). Machine learning can find the subtle, multi-variable patterns that indicate a page is about to lose traffic, allowing us to act proactively rather than reactively.
